### Question 1: Import Data and Store-wise Revenue Analysis (4 Marks)

Import `pandas` and `numpy`. Download the dataset `sales.csv` from https://drive.google.com/drive/folders/1gieHICVDBbUKMZiSF4YRQMAJic-w50JM?usp=sharing. Load the dataset `sales.csv` into a DataFrame. Then, find the top 5 stores with the highest total sales and the bottom 5 stores with the lowest total sales.

**Tasks:**
- Aggregate total sales by store_id
- Rank stores by total revenue
- Compare average sales per department within those stores

```python
# Write your answer here
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv('sales.csv')

# --- Task 1: Aggregate total sales by store_id ---
store_revenue = df.groupby('store_id')['weekly_sales'].sum().reset_index()

# --- Task 2: Rank stores by total revenue ---
# Sorting in descending order to rank from highest to lowest revenue
store_revenue_sorted = store_revenue.sort_values(by='weekly_sales', ascending=False).reset_index(drop=True)
store_revenue_sorted['rank'] = store_revenue_sorted['weekly_sales'].rank(ascending=False, method='min').astype(int)

# Extract Top 5 and Bottom 5 stores
top_5_stores_df = store_revenue_sorted.head(5)
bottom_5_stores_df = store_revenue_sorted.tail(5)

print("--- Top 5 Stores with Highest Total Sales ---")
print(top_5_stores_df.to_string(index=False))

print("\n--- Bottom 5 Stores with Lowest Total Sales ---")
print(bottom_5_stores_df.to_string(index=False))

# --- Task 3: Compare average sales per department within those stores ---
top_5_ids = top_5_stores_df['store_id'].tolist()
bottom_5_ids = bottom_5_stores_df['store_id'].tolist()
selected_stores = top_5_ids + bottom_5_ids

# Filter original data for only these 10 stores
filtered_df = df[df['store_id'].isin(selected_stores)]

# Calculate average sales per department for each selected store
dept_avg_sales = filtered_df.groupby(['store_id', 'department'])['weekly_sales'].mean().reset_index()

# Pivot the table to create a clear side-by-side comparison across stores
comparison_pivot = dept_avg_sales.pivot(index='department', columns='store_id', values='weekly_sales')

print("\n--- Department Average Sales Comparison Matrix (First 5 Rows) ---")
print(comparison_pivot.head(5))

# Export the detailed matrix to a CSV file
comparison_pivot.to_csv('department_sales_comparison.csv')
```

### Question 2: Department Performance Consistency (3 Marks)

Which departments have the most stable weekly sales and which are the most volatile?

**Tasks:**
- Use standard deviation and coefficient of variation
- Identify highly stable and highly volatile departments

```python
# Write your answer here
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv('sales.csv')

# Task 01:- Calculate Mean, Standard Deviation, and Coefficient of Variation.
# Group by department to aggregate metrics
dept_metrics = df.groupby('department')['weekly_sales'].agg(['mean', 'std']).reset_index()

# Calculate Coefficient of Variation (CV)
dept_metrics['cv'] = dept_metrics['std'] / dept_metrics['mean']


# Task 02:- Identify Highly Stable and Highly Volatile Departments.
# Stable departments have the lowest CV values
stable_departments = dept_metrics.sort_values(by='cv', ascending=True).head(5)

# Volatile departments have the highest CV values
volatile_departments = dept_metrics.sort_values(by='cv', ascending=False).head(5)


print("--- Top 5 Most Stable Departments (Lowest CV) ---")
print(stable_departments.to_string(index=False, formatters={
    'mean': '${:,.2f}'.format,
    'std': '${:,.2f}'.format,
    'cv': '{:.4f}'.format
}))

print("\n--- Top 5 Most Volatile Departments (Highest CV) ---")
print(volatile_departments.to_string(index=False, formatters={
    'mean': '${:,.2f}'.format,
    'std': '${:,.2f}'.format,
    'cv': '{:.4f}'.format
}))
```

### Question 3: Holiday vs Non-Holiday Sales (3 Marks)

Analyze whether holidays significantly affect weekly sales.

**Tasks:**
- Compare average sales during holidays vs non-holidays
- Calculate percentage increase/decrease
- Identify departments benefiting most from holidays

```python
# Write your answer here
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv('sales.csv')

# Task 01:- Compare average sales during holidays vs non-holidays.
holiday_summary = df.groupby('is_holiday')['weekly_sales'].mean().reset_index()
holiday_summary['is_holiday_label'] = holiday_summary['is_holiday'].map({0: 'Non-Holiday', 1: 'Holiday'})

avg_non_holiday = holiday_summary.loc[holiday_summary['is_holiday'] == 0, 'weekly_sales'].values[0]
avg_holiday = holiday_summary.loc[holiday_summary['is_holiday'] == 1, 'weekly_sales'].values[0]

print("--- Task 1: Average Weekly Sales Comparison ---")
print(f"Average Sales (Non-Holiday Weeks): ${avg_non_holiday:,.2f}")
print(f"Average Sales (Holiday Weeks):     ${avg_holiday:,.2f}")


# Task 02:- Calculate percentage increase/decrease.
pct_increase = ((avg_holiday - avg_non_holiday) / avg_non_holiday) * 100
print("\n--- Task 2: Overall Holiday Percentage Impact ---")
print(f"Holidays drive a significant increase of {pct_increase:.2f}% in overall average weekly sales.")


# Task 03:- Identify departments benefiting most from holidays.
# Group by department and holiday status to isolate average performance matrices
dept_perf = df.groupby(['department', 'is_holiday'])['weekly_sales'].mean().unstack().reset_index()
dept_perf.columns = ['department', 'avg_non_holiday', 'avg_holiday']

# Compute both percentage relative gains and absolute dollar value gains
dept_perf['pct_growth'] = ((dept_perf['avg_holiday'] - dept_perf['avg_non_holiday']) / dept_perf['avg_non_holiday']) * 100
dept_perf['absolute_growth'] = dept_perf['avg_holiday'] - dept_perf['avg_non_holiday']

# Rank and isolate top 5 departments
top_5_pct_growth = dept_perf.sort_values(by='pct_growth', ascending=False).head(5)

print("\n--- Task 3: Top 5 Departments Benefiting Most (by % Growth) ---")
print(top_5_pct_growth.to_string(index=False, formatters={
    'avg_non_holiday': '${:,.2f}'.format,
    'avg_holiday': '${:,.2f}'.format,
    'pct_growth': '{:.2f}%'.format,
    'absolute_growth': '${:,.2f}'.format
}))
```
